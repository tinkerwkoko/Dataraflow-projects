- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit must be positive.")
            return
        self.balance += amount
        print(f"Deposited {amount}. Balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
            return
        self.balance -= amount
        print(f"Withdrew {amount}. Balance: {self.balance}")


class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, interest_rate=0.05):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Interest added: {interest}. Balance: {self.balance}")


account = SavingsAccount("Annie", 1000, 0.05)
account.deposit(500)
account.withdraw(200)
account.withdraw(5000)
account.add_interest()
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    def __init__(self, n):
        self.n = n

    def __iter__(self):
        self.current = self.n
        return self

    def __next__(self):
        if self.current < 1:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value


for num in CountDown(5):
    print(num)
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
class FileLineReader:
    def __init__(self, filename):
        self.filename = filename

    def __iter__(self):
        self.file = open(self.filename, "r")
        return self

    def __next__(self):
        line = self.file.readline()
        if line == "":
            self.file.close()
            raise StopIteration
        return line.rstrip("\n")


# Create a sample file to test with
with open("sample.txt", "w") as f:
    f.write("First line\nSecond line\nThird line\n")

for line in FileLineReader("sample.txt"):
    print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
from abc import ABC, abstractmethod
import math


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


shapes = [Rectangle(4, 5), Circle(3), Rectangle(2, 7)]

for shape in shapes:
    print(f"Area: {shape.area():.2f}")
````

````python
def show_len_behavior():
    numbers = [1, 2, 3]

    print(f"Built-in len(): {len(numbers)}")

    def shadow():
        len = 100
        try:
            len(numbers)
        except TypeError as e:
            print(f"Error: {e}")

    shadow()

show_len_behavior()
````

# inside shadow() creates a local name that is found first, so it hides the
# built-in. Calling it then fails because 100 is an integer, not a function.
# The built-in len() is not changed, and it works normally everywhere the
# local variable does not exist.

Create two classes: `Flyer` (method `move()` returns `"Flying"`) and `Swimmer` (method `move()` returns `"Swimming"`).  
Create a class `Duck` inheriting from both.  
- Use `super()` to decide which `move()` executes.  
- Explain the **Method Resolution Order (MRO)**.  

````python
class Flyer:
    def move(self):
        return "Flying"


class Swimmer:
    def move(self):
        return "Swimming"


class Duck(Flyer, Swimmer):
    def move(self):
        return super().move()


duck = Duck()
print(duck.move())
print(Duck.__mro__)
````

# The Method Resolution Order (MRO) is the order in which Python searches
# classes when looking up a method or attribute. Python builds it using the
# C3 linearization algorithm, which keeps each class before its parents and
# preserves the left-to-right order of the base classes.
# Duck(Flyer, Swimmer) gives the order Duck, Flyer, Swimmer, object.
# super().move() inside Duck starts searching after Duck in that order, so
# it finds Flyer.move() first and returns "Flying". Swimmer.move() is never
# reached because Flyer comes earlier. Swapping the base classes to
# Duck(Swimmer, Flyer) would make the result "Swimming". You can inspect
# the order at any time with Duck.__mro__ or Duck.mro().

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
    def __init__(self, n):
        self.n = n

    def __iter__(self):
        self.a = 0
        self.b = 1
        return self

    def __next__(self):
        if self.a > self.n:
            raise StopIteration
        value = self.a
        self.a, self.b = self.b, self.a + self.b
        return value


for num in Fibonacci(50):
    print(num)
````

## Scenario

### 5. Extra Features  
- Track which books are available and which are borrowed.

````python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_available = True

    def get_description(self):
        return f"'{self.title}' by {self.author} ({self.year})"


class EBook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        return f"{super().get_description()} | E-book: {self.file_size} MB, {self.format}"


class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return f"{super().get_description()} | Printed: {self.weight} kg, {self.dimensions}"


class Member:
    total_borrowed = 0

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = []  # protected

    def get_borrowed_books(self):
        return list(self._borrowed_books)

    def borrow_limit(self):
        return 1

    def borrow_book(self, book):
        if not book.is_available:
            print(f"'{book.title}' is not available.")
            return False
        if len(self._borrowed_books) >= self.borrow_limit():
            print(f"{self.name} has reached the borrowing limit of {self.borrow_limit()} books.")
            return False
        self._borrowed_books.append(book)
        book.is_available = False
        Member.total_borrowed += 1
        print(f"{self.name} borrowed '{book.title}'.")
        return True

    def return_book(self, book):
        if book in self._borrowed_books:
            self._borrowed_books.remove(book)
            book.is_available = True
            print(f"{self.name} returned '{book.title}'.")
        else:
            print(f"{self.name} has not borrowed '{book.title}'.")

    # Iterator over borrowed books
    def __iter__(self):
        self._index = 0
        return self

    def __next__(self):
        if self._index >= len(self._borrowed_books):
            raise StopIteration
        book = self._borrowed_books[self._index]
        self._index += 1
        return book


class StudentMember(Member):
    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level

    def borrow_limit(self):
        return 2


class TeacherMember(Member):
    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department

    def borrow_limit(self):
        return 4


class Library:
    def __init__(self, name):
        self.__library_name = name
        self.books = []
        self.members = []

    def get_library_name(self):
        return self.__library_name

    def set_library_name(self, new_name):
        if not new_name.strip():
            raise ValueError("Library name cannot be empty.")
        self.__library_name = new_name

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def available_books(self):
        return [b for b in self.books if b.is_available]

    def borrowed_books(self):
        return [b for b in self.books if not b.is_available]

    def search(self, keyword):
        keyword = keyword.lower()
        return [b for b in self.books
                if keyword in b.title.lower() or keyword in b.author.lower()]

    # Iterator over all books
    def __iter__(self):
        self._index = 0
        return self

    def __next__(self):
        if self._index >= len(self.books):
            raise StopIteration
        book = self.books[self._index]
        self._index += 1
        return book


library = Library("City Central Library")

ebook1 = EBook("Python Fundamentals", "Ada Okoye", 2021, 4.5, "PDF")
ebook2 = EBook("Data Structures Explained", "Chidi Eze", 2019, 8.2, "EPUB")
print1 = PrintedBook("Algorithms in Practice", "Ngozi Bello", 2018, 1.2, "24x18x3 cm")
print2 = PrintedBook("Database Design", "Ada Okoye", 2022, 0.9, "22x15x2 cm")

for book in (ebook1, ebook2, print1, print2):
    library.add_book(book)

student = StudentMember("Tunde", "M001", "300 Level")
teacher = TeacherMember("Mrs. Adeyemi", "M002", "Computer Science")
library.add_member(student)
library.add_member(teacher)

print(f"=== {library.get_library_name()} ===")

print("\n--- All books (Library iterator) ---")
for book in library:
    print(book.get_description())

print("\n--- Members ---")
print(f"{student.name} ({student.grade_level}), ID {student.member_id}")
print(f"{teacher.name} ({teacher.department}), ID {teacher.member_id}")

print("\n--- Borrowing ---")
student.borrow_book(ebook1)
student.borrow_book(print1)
student.borrow_book(ebook2)
teacher.borrow_book(ebook1)
teacher.borrow_book(ebook2)
teacher.borrow_book(print2)

print("\n--- Tunde's borrowed books (Member iterator) ---")
for book in student:
    print(book.title)

print(f"\nTotal books borrowed by all members: {Member.total_borrowed}")

print("\n--- After Tunde returns a book ---")
student.return_book(ebook1)
print("Available:", ", ".join(b.title for b in library.available_books()))
print("Borrowed:", ", ".join(b.title for b in library.borrowed_books()))
print(f"Total borrowed (cumulative): {Member.total_borrowed}")

print("\n--- Search ---")
print("Search 'okoye':", ", ".join(b.title for b in library.search("okoye")))
print("Search 'algorithms':", ", ".join(b.title for b in library.search("algorithms")))

print("\n--- Scope ---")
print("Library name:", library.get_library_name())
library.set_library_name("Lagos City Library")
print("Updated name:", library.get_library_name())
try:
    print(library.__library_name)
except AttributeError as e:
    print(f"Direct access failed: {e}")
print("Tunde's borrowed titles via getter:", [b.title for b in student.get_borrowed_books()])
````
